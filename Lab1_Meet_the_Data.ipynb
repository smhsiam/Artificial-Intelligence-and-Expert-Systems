**Key words**
- **Machine Learning:** teaching a computer with examples instead of rules.
- **Dataset:** a table of examples (like an Excel sheet).
- **Feature (X):** the input the model looks at - here the **Text** column.
- **Label (y):** the answer the model must guess - here the **Emotion** column.
- **Classification:** the answer is a category (joy, anger...), not a number.
